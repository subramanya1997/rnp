## Create typed arrays

Build a two-dimensional integer grid and an evenly spaced floating-point vector.

```python
grid = np.arange(12, dtype=np.int64).reshape(3, 4)
unit_interval = np.linspace(0.0, 1.0, 5, dtype=np.float64)
print("grid shape/dtype:", grid.shape, grid.dtype)
print(grid)
print("unit interval:", unit_interval)
```

## Index and broadcast

Compare basic slicing with paired fancy indices, then broadcast a row across two columns.

```python
sliced = grid[::2, 1::2]
fancy = grid[[2, 0], [1, 3]]
broadcast_sum = grid[:, :2] + np.array([10, 100], dtype=np.int64)
print("slice:", sliced)
print("fancy index:", fancy)
print("broadcast result:\n", broadcast_sum)
```

## Verify the result

```python
assert np.array_equal(sliced, [[1, 3], [9, 11]])
assert np.array_equal(fancy, [9, 3])
assert np.array_equal(broadcast_sum[-1], [18, 109])
assert np.array_equal(unit_interval, [0.0, 0.25, 0.5, 0.75, 1.0])
print("PASS — all array-basics assertions passed.")
```
